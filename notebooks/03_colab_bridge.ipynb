**Pregunta:** ¿Qué tienda tiene el mayor promedio de ventas diarias en una semana?

**Procedencia de los datos:** Datos **sintéticos** generados dentro de este notebook con NumPy (semilla fija `42`). No se usan datos reales, personales ni confidenciales, y no se monta Google Drive.

**Resultado esperado:** Una tabla con el promedio de ventas diarias por tienda y el nombre de la tienda con el promedio más alto.

**Dependencias:** Python 3, `numpy`, `pandas` (ya vienen en Colab; si no, la celda de setup las instala).

```python
# Setup: versión de Python y paquetes necesarios
import sys
print("Python:", sys.version)

# Instala solo si faltan (Colab ya las trae)
import importlib.util, subprocess
for pkg in ["numpy", "pandas"]:
    if importlib.util.find_spec(pkg) is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])

import numpy as np
import pandas as pd
print("numpy:", np.__version__)
print("pandas:", pd.__version__)
```

```python
# Datos sintéticos: 3 tiendas x 7 días
rng = np.random.default_rng(42)
tiendas = ["Tienda A", "Tienda B", "Tienda C"]
dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]

df = pd.DataFrame(
    [(t, d, int(rng.integers(800, 2000))) for t in tiendas for d in dias],
    columns=["tienda", "dia", "ventas"],
)
df.head(10)
```

```python
# Análisis: promedio de ventas diarias por tienda
resumen = (
    df.groupby("tienda")["ventas"]
      .mean()
      .round(1)
      .sort_values(ascending=False)
      .rename("promedio_ventas_diarias")
      .reset_index()
)
top = resumen.iloc[0]
print(f"Tienda con mayor promedio: {top['tienda']} ({top['promedio_ventas_diarias']})")
resumen
```

La tabla anterior muestra el promedio de ventas diarias por tienda y cuál tiene el promedio más alto. Como los datos son sintéticos con semilla fija, cualquier persona que corra este notebook desde un runtime limpio obtiene el mismo resultado.
